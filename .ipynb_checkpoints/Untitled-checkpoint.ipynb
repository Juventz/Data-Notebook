# Transport Data Analysis
Explore, analysis and SQL modelling based on a csv file of public transport journeys

**Objectives:**
- Data cleansing
- Calculation of KPIs (punctuality, delays, ridership)
- Creation of a SQLModel with insertion asn SQLite queries

## Library import

```python
import pandas as pd
from sqlmodel import *
import sqlite3
from datetime import datetime
```

## CSV file load

## Data Cleansing

## Convert dates

```python
df['planned_departure'] = pd.to_datetime(df['planned_departure'], errors='coerce')
df['actual_arrival'] = pd.to_datetime(df['actual_arrival'], errors='coerce')
```

## Deleting undefined, missing values and duplicates

```python
df.isnull().sum()
```

```python
df.isna().sum()
```

```python
df.dropna()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['duration_minutes'] = (df['actual_arrival'] - df['planned_departure']).dt.total_seconds() / 60
```

```python
print(df)
```

# Data Analysis

## Kpis : on_time, mean_delay, longest_trip, shortest_trip, most_used_line, num_trips.

```python
on_time  = df[df['delay_minutes'] <= 5]
print('Taux de ponctualité :', len(on_time) / len(df) * 100, "%")
```

```python
mean_delay = df['delay_minutes'].mean()
print('Retard moyen :', mean_delay, "%")
```

```python
longest_trip = df.loc[df['duration_minutes'].idxmax()]
print("Longest trip :", longest_trip)
```

```python
shortest_trip = df.loc[df['duration_minutes'].idxmin()]
print('Shortest trip :', shortest_trip)
```

```python
most_used_line = df.groupby('line')['passenger_count'].sum().idxmax()
print('Most used line :', most_used_line)
```

```python
num_trips = df[df['line'] == most_used_line].shape[0]
print(f"Number of trips for the  {most_used_line} : {num_trips}")
```

# MODEL SQL

model creation with ORM sqlmodel

```python
class Trip(SQLModel, table=True):
    trip_id: int = Field(primary_key=True)
    line: str
    departure_stop: str
    arrival_stop: str
    planned_departure: datetime
    actual_arrival: datetime
    passenger_count: int
    delay_minutes: int
    duration_minutes: float
```

```python
engine = create_engine("sqlite:///transport.db")
SQLModel.metadata.create_all(engine)

with Session(engine) as session:
    for row in df.itertuples(index=False):
        trip = Trip(**row._asdict())
        session.add(trip)
    session.commit()
```

Request avec SQLModel

```python
with Session(engine) as session:
    statement = select(Trip.line, func.avg(Trip.delay_minutes).label('avg_delay')).group_by(Trip.line)
results = session.exec(statement).all()
print(results)
```

Retour dans pandas via Requete SQL

```python
df_sql = pd.read_sql_query("SELECT line, AVG(delay_minutes) as avg_delay FROM Trip Group BY line", engine)
print(df_sql)
```
